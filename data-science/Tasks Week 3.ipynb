# Data Science Track: Week 3

With the article clusters established from Week 2, we now transition to predictive modeling. In retail, accurate demand forecasting directly impacts write-offs and availability and therefore customer satisfaction.

At Kaufland, fruit and vegetables are mostly ordered from central distribution centers (DC). However, before the store can order articles from the DC, the DC has to order them from the suppliers first. Depending on the article, this can be more than a week before the store orders even arrive. In this sprint, we will mirror those lead times by creating an aggregated demand forecasting model for different horizons.

For this sprint, your objective is to build a **univariate** demand forecasting workflow to predict aggregated sales for **T+2** and **T+7** horizons across selected articles. You will establish performance baselines, log experiments using **MLflow**, train supervised regression models, and evaluate performance using industry-standard retail metrics.

**Hint:** Univariate Forecasting excludes exogenous features. We will include these in Week 4.

**Suggested Tools:** `pandas`, `numpy`, `scikit-learn`, `mlflow`, `streamlit`

**Hint:** Used features must be available at prediction time. Be wary of temporal leakage.

## Part 2: Metrics Definition & Baseline Models

To evaluate performance realistically, we must define domain-specific retail metrics and establish simple baseline models.

- Implement functions to compute the Weighted Mean Absolute Percentage Error (**WMAPE**) and **BIAS**. 

$$\text{WMAPE} = \frac{\sum |Actual - Forecast|}{\sum |Actual|}$$

$$\text{BIAS} = \frac{\sum (Forecast - Actual)}{\sum (Forecast + Actual)}$$

- In a Markdown cell, document how you would explain WMAPE and BIAS to non-technical business stakeholders.
- Research and implement at least **two** simple time-series baselines. Justify your choice.
- Use **MLflow** to log parameters, metrics (WMAPE and BIAS) and baseline predictions for the test period (Business year 2025).
