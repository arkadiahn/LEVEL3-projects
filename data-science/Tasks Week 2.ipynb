In Week 1, we built a unified `.parquet` dataset. This week, we'll be using unsupervised clustering, a machine learning technique that discovers natural groupings in data without needing target labels to divide the articles and stores into different groups.

### Why do we cluster?

In retail forecasting, treating every item identically is a recipe for failure:

* **Divergent Sales Behavior:** A banana behaves nothing like a strawberry. Bananas are year-round staples, while strawberries experience seasonal demand spikes.
* **The Scale Problem:** Kaufland manages thousands of article-store combinations. Building a custom forecasting logic or inventory strategies for every individual article can be challenging. Clustering reduces thousands of articles into a few distinct behavioral archetypes.
* **Value for Modeling:** Grouping products or stores by their sales dynamics allows us to build smarter downstream forecasting models by, for example, training separate model architectures for stable items versus seasonal items.
