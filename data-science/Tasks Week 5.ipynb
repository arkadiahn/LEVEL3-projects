# Data Science Track: Week 5

Your forecasting model from Week 4 performs well on historical data, but a model has to be maintained constantly. In the real world, consumer behavior changes, macroeconomic factors shift, and static models decay. Furthermore, business users need an accessible way to interact with your predictions.

For this final sprint, your objective is twofold: **Deployment** and **Monitoring**. You will finalize your Streamlit dashboard to serve live predictions, simulate production via historical backtesting, implement drift detection, and define a model retraining strategy.

**Suggested Tools:** `pandas`, `numpy`, `scikit-learn`, `mlflow`, `skforecast`, `streamlit`

## Part 1: Interactive Forecasting Dashboard

Until now, your dashboard has displayed static, pre-calculated predictions. It is time to upgrade your application to serve live predictions on demand.

- Save your best performing models from Week 4.
- Add a new view or tab to your existing `app.py`.
- **App Requirements:**
    - The dashboard must load the pre-trained models and generate predictions **on-the-fly** based on user interaction (e.g., selecting a store-article combination and predicting the next days).
    - Provide visualizations plotting *Predicted Sales* vs. *Actual Sales*.
    - Include UI controls allowing users to select specific stores and articles.
    - Dynamically calculate and display the **WMAPE** and **BIAS** for the currently selected data slice.
    - Provide a "Download" button allowing the user to export the currently selected data as a `.csv` file.

## Part 2: Drift Detection Strategy

Establish a monitoring strategy to detect shifts in the input data.

- Implement a **statistical strategy** in code to detect data drift between two datasets. Justify your decision.
- Integrate your drift detection checks into your Streamlit dashboard to visually alert users of shifts.

## Part 3: Historical Backtesting & Retraining Plan

Continuous retraining is expensive. The engineering team needs you to define exactly how often the model should be retrained and with how much historical data.

Take the feature engineering logic and model architecture you selected in Week 4. Instead of using a static pre-trained model, use your pipeline to run a backtesting simulation on your selected article-store data.

- Implement historical backtesting to simulate how your model would have performed in production over time.
- Evaluate different retraining strategies, like static models (trained once) or dynamic models that retrain on a rolling or expanding window.
- Test various configurations by varying the *retraining frequency* (e.g., weekly vs. monthly) and the *training data window* (e.g., 3 months vs. 12 months). Track your experiments with MLflow.
- Based on the trade-off between metric improvement (WMAPE/BIAS) and operational overhead, formulate and document a final recommendation for the deployment team.

**Hint:** https://skforecast.org/latest/user_guides/backtesting.html

## Bonus: Crisis Simulation — Handling Real-World Concept Drift

**Scenario:** You have been provided with a new test dataset spanning the same period, but representing a completely different retail environment. A major market shift has occurred.

- Run your drift detection strategy from Part 2 on this new dataset. Quantify the severity of the drift.
- Evaluate your current model's performance on this new dataset to measure the model decay.
- Think of techniques to adjust your model to this new reality. Describe your proposed solution.
- Implement your mitigation strategy and prove that it improves the WMAPE/BIAS metrics compared to the unadjusted model.
