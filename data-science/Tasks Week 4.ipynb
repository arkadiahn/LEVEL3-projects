# Data Science Track: Week 4

In Week 3, we forecasted the aggregated demand, so the Distribution Center (DC) could order the required quantities from suppliers. Now that sufficient stock is available at the DC, individual stores want to order based on their own local customer demand.

For this sprint, your objective is to transition from **univariate** to **multivariate forecasting** at the more granular store-article level. You will engineer exogenous features, apply feature selection, train local models for the `T+3` horizon and extract business insights using feature importance frameworks.

**Suggested Tools:** `pandas`, `numpy`, `scikit-learn`, `mlflow`, `shap`, `streamlit`

## Part 1: Feature Engineering & Selection

To predict sales accurately, you must extract meaningful signals from the data beyond just historical target values. Standard machine learning algorithms require numerical values, meaning categorical and temporal data must be explicitly transformed.

- Engineer a diverse set of features.
- Implement a methodology for feature selection.
- Explain your chosen selection method and document which features were retained and dropped.

**Hint:** Categorical and temporal features must be encoded appropriately. Used features must be available up to 3 days into the future. Be wary of temporal leakage.

## Part 2: Local Forecasting (Store-Article Level)

You will now train **local models** predicting exactly three days into the future (`T+3`).

- Select a suitable forecasting model of your choice. Justify your choice.
- Train separate forecasting models for **15 distinct combinations** (3 articles * 5 stores). 
- **Mandatory Combinations:** 
    - **Bananas** and **Oranges** in stores 2350, 2250, 2850, 1000 and 1900.
    - Please forecast one more article in five stores of your choice.
- As in Week 3, establish a baseline for these combinations and evaluate your models using WMAPE and BIAS. Your trained models must outperform the baselines and be logged in MLflow.

**Hint:** Think about which article behavior would be interesting to predict. Be creative: try to experiment with different features and reach the best possible metrics. The more you experiment and research, the more you will learn.

## Part 3: Explainability & Business Impact

Even an accurate model is often doubted by retail planners if it operates as a black box. You must explain how the model makes its predictions.

- Compute and visualize SHAP values for one of your store-article models.
- Write a summary interpreting the primary drivers of demand for that specific store-article combination. Be wary of the limitations of this method.

## Part 4: Dashboard Forecast Visualization
The stakeholders want to see how your models perform visually.

- Save the test set actuals and `T+3` predictions for your 15 combinations to a file.
- Add a new view or tab to your existing `app.py`.
- **App Requirements:**
    - Provide UI controls allowing the user to select both an **Article** and a **Store**.
    - Display a suitable visualization comparing Actual Sales vs. the `T+3` predictions over the test period for the selected combination.
    - Dynamically calculate and display the WMAPE and BIAS for the selection.
    - Integrate a visualization of the Feature Importance to explain the model's behavior to the user.
    - Add a download button allowing the user to export the selected forecast data as a `.csv` file.
    - Feel free to add more things you think are interesting.

Always keep usability and your stakeholders in mind.

## Bonus: Global Modeling Strategy

Training thousands of local models (one for every store-article combination) is computationally expensive and sometimes difficult to maintain. 

- Utilize the article clusters you identified in Week 2 to train a single **global model** across multiple articles and stores. Log it in MLflow.
- Adjust your feature engineering pipeline to ensure the global model can distinguish between different stores and articles.
- Evaluate the global model's performance against the local models. Summarize the trade-offs of this approach.
- Integrate the results into your Dashboard by exporting the predictions of your global model and adding a visualization to compare the default model's forecast against the global model's forecast.
